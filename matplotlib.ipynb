```python
import matplotlib.pyplot as plt
```

```python
# line graph
students = ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"]
marks = [70, 90, 75, 85, 60]
plt.plot(students, marks)
plt.xlabel("students")
plt.ylabel("marks")
plt.show()
```

```python
# Bar chart
plt.bar(students,marks)
plt.xlabel("students")
plt.ylabel("marks")
plt.show()
```

```python
# bar horizontal graph
plt.barh(students,marks)
plt.xlabel("marks")
plt.ylabel("students")
plt.show()
```

```python
# scatter plot
study_hours = [2,3,4,5,6,7]
marks = [70, 90, 75, 85, 60, 50]
plt.scatter(study_hours,marks)
plt.xlabel("study hours")
plt.ylabel("marks")
plt.show()
```

```python
# scatter plot example # 2 experience vs salary
experience = [5,4,2,3,6,7,8]
salary = [50000,30000,70000,90000,10000,10000,20000]
plt.scatter(experience,salary)
plt.xlabel("experience")
plt.ylabel("salary")
plt.show()
```

```python
# histogram plot
plt.hist(salary,bins=5)
plt.xlabel("salary")
plt.ylabel("frequency")
plt.show()
```

```python
# pie chart
categories = ['Rent', 'Food', 'Utilities', 'Entertainment',"Rent"]
percentages = [45, 25, 15, 15,45]
plt.pie(percentages, labels=categories.value_count.index, autopct='%1.1f%%')
plt.show()
```

```python
# task # 1
Months = ["Jan", "Feb", "Mar", "Apr", "May"]
Sales  = [20, 35, 30, 50, 45]
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.plot(Months,Sales)
plt.show()
```

```python
# task # 2
courses = ["python","AI","Web","Design"]
mark = [40,55,35,25]
plt.title("Courses")
plt.xlabel("courses")
plt.ylabel("mark")
plt.bar(courses,mark)
plt.show()
```
